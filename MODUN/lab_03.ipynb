# Równianie Poissona
- **$\textbf{M} = \textbf{D} + \textbf{L} + \textbf{U}$**

**Jacobi:**
- **$\textbf{v}_{k+1} = -\textbf{D}^{-1}[\rho + (\textbf{L} + \textbf{U})\textbf{v}_k]$**
  
- **$\textbf{v}^{k+1}_i = -\textbf{M}^{-1}[\rho_i + (\sum_{i \neq j}\textbf{M}_{ij})\textbf{v}^k_i]$**

**G-S:**
- **$\textbf{v}_{k+1} = -(\textbf{D} + \textbf{L})^{-1}[\rho + \textbf{Uv}_k]$**

- **$\textbf{v}_{k+1} = -(\textbf{D})^{-1}[\rho + \textbf{L}\textbf{v}_{k+1} + \textbf{Uv}_k]$**
- **$\textbf{v}^{k+1}_p = -(\textbf{M}_{pp})^{-1}[\frac{\rho}{\epsilon} + \sum_{q<p}\textbf{M}_{pq}\textbf{v}^{k+1}_q + \sum_{q<p}\textbf{M}_{pq}\\\textbf{v}_k]$**
