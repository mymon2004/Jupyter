```python
#INTEGER NUMBERS IN VECTORS
tab1 = np.zeros(2) #vector with only zeroes
print(tab1)
tab1[1] = 4 #numeration starts from 0
print(tab1)
tab2 = np.arange(1, 4, 1) 
print(tab2)
tab3 = np.arange(1, 5, 1) #1 - first value, 5 - last value (but actually it's smaller?), 1 - jump with +1
tab4 = np.arange(1, 40, 5) #1 - first value, 40 - last value (but actually it's smaller?), 5 -jump with +5
print(tab3)
    
print(tab4)
   
tab5 = np.zeros(4)
tab5[2] = tab3[0] + tab4[0]
print(tab5)
```

```python
d = 1
for i in range(10):
    d = d * np.exp(3)
    print(d) #body of the loop
    
def func(x): #defining a function
    return np.exp(x**2) + np.sin(x) 
    
xtab = np.linspace(0.0, 2.0, 100) #the range of x 
ytab = np.zeros(100) #values of function
for i in range(100):
    ytab[i]=func(xtab[i])
plt.plot(xtab, ytab)
```
