```python
import numpy as np
import matplotlib.pyplot as plt

from scipy.constants import physical_constants
```

```python
h_bar = physical_constants['Planck constant over 2 pi'][0]
m_el = physical_constants['electron mass'][0]
eV_unit = physical_constants['electron volt'][0]

a = 1.0 * 10**(-9)            #m  
L = 50*a

N = int(L/a) - 1

t_unit_J = h_bar**(2)/(2*m_el*a**2)  #in the units of J 
t = t_unit_J/eV_unit                 #in the units of eV

print("a [m]:", a)
print("t [J]:", t_unit_J)
print("t [eV]:", t)
print("L =", L)
print("N =", N)
print("L(N+1) =", a*(N+1))
```
