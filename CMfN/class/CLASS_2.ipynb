```python
import numpy as np
import matplotlib.pyplot as plt

from scipy.constants import physical_constants
```

```python
a = 1.0 * 10**(-9)
N = 49
L = (N+1)*a
print("L=",L,"m")
print("L=",L/a,"nm")
```

```python
h_bar = physical_constants['Planck constant over 2 pi'][0]
eV_unit = physical_constants['electron volt'][0]
m_el = physical_constants['electron mass'][0]
```

```python
t = (h_bar**2) / (2*m_el*a**2)
t = t / eV_unit
print("t=",t)
```

```python
Hmat = np.zeros((N,N))
```

```python
for i in range(N):
    Hmat[i][i] = 2*t
    if(i<N-1):
        Hmat[i][i+1] = -t
        Hmat[i+1][i] = -t
```

```python
print(Hmat/t)
```

```python
ene, v = np.linalg.eigh(Hmat)
```

```python
ntab = np.arange(N)
plt.plot(ntab, ene, "r.");
plt.xlabel("n")
plt.ylabel("E [eV]")
plt.title("Energy ")
```
