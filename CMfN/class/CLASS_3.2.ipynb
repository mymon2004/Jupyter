```python
# QUANTUM WELL CIRCLE
import numpy as np
import matplotlib.pyplot as plt
import kwant

from scipy.constants import physical_constants
```

```python
a = 0.5 * 10**(-9)
a_nm = a * 10**9
N = 40
L = (N+1)*a
L_nm = L*10**9
print("L=",L,"m")
print("L=",L/10**(-9),"nm")
```

```python
h_bar = physical_constants['Planck constant over 2 pi'][0]
eV_unit = physical_constants['electron volt'][0]
m_el = physical_constants['electron mass'][0]
```

```python
t = (h_bar**2) / (2*m_el*a**2)
t = t / eV_unit
print("t=",t)
```

```python
sys = kwant.Builder()
```

```python
lat = kwant.lattice.square(a_nm)
```

```python
R = 15
def shape1(pos):
    (x,y) = pos
    if x**2 + y**2 < R**2:
        return True
    else:
        return False
    
    
sys[lat.shape(shape1, (0,0))] = 4*t
sys[lat.neighbors()] = -t

kwant.plot(sys);
```

```python
sysf = sys.finalized()
```

```python
Hmat = sysf.hamiltonian_submatrix()
```

```python
ene, v = np.linalg.eigh(Hmat)
```

```python
plt.plot(ene[:10], "r.");
```

```python
from ipywidgets import interact
```

```python
def plot_wf(n=0):
    kwant.plotter.map(sysf, abs(v[:,n])**2);
```

```python
interact(plot_wf, n=(0,20))
```
